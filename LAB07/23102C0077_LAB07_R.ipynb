# Lecture 30 – Lists

### Matrices vs Lists behavior: Matrix addition

```r
x1 = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = TRUE)
x2 = matrix(nrow = 2, ncol = 2, data = 5:8, byrow = TRUE)
x1
x2
x1 + x2
```

### Matrix element modification causing type conversion and binary addition error (Handled safely)

```r
x1[2, 1] = "hello"
x1
# Executing x1 + x2 safely using tryCatch to demonstrate expected error
tryCatch({
  x1 + x2
}, error = function(e) {
  print(e)
})
```

### Creating lists containing matrices

```r
x1 = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = TRUE)
x2 = matrix(nrow = 2, ncol = 2, data = 5:8, byrow = TRUE)
x1
x2
```

```r
matlist = list(x1, x2)
matlist
```

### Extracting elements / sublists from a list using []

```r
matlist[1]
```

```r
matlist[2]
```

### List containing different object types

```r
z1 = list(c("water", "juice", "lemonade"), rep(1:4, each = 2), matrix(data = 5:8, nrow = 2, ncol = 2, byrow = TRUE))
z1
```
