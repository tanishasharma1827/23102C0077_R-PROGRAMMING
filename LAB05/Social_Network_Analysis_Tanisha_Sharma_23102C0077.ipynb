```r
# Cell 5 — Create and Prepare Network Dataset
edge_data <- data.frame(
  from = c(
    "Alice", "Alice", "Alice", "Alice",
    "Bob", "Bob",
    "Charlie", "Charlie",
    "David", "David",
    "Emma", "Emma", "Emma",
    "Frank", "Frank",
    "Grace", "Grace",
    "Hannah",
    "Ian"
  ),
  to = c(
    "Bob", "Charlie", "David", "Emma",
    "Charlie", "David",
    "David", "Frank",
    "Frank", "Grace",
    "Frank", "Grace", "Hannah",
    "Grace", "Hannah",
    "Hannah", "Ian",
    "Ian",
    "Jack"
  ),
  weight = c(4, 5, 2, 3, 3, 2, 4, 3, 2, 1, 4, 2, 5, 3, 2, 3, 4, 4, 2),
  stringsAsFactors = FALSE
)

cat("First 8 interactions in the edge list:\n")
print(head(edge_data, 8))
cat("\nTotal interactions (edges):", nrow(edge_data), "\n")
```

```r
# Cell 6 — Construct Graph Object in R
net <- graph_from_data_frame(d = edge_data, directed = FALSE)

cat("Is the graph directed?", is_directed(net), "\n")
cat("Total number of vertices (nodes):", vcount(net), "\n")
cat("Total number of links (edges):", ecount(net), "\n")
cat("Graph density:", edge_density(net), "\n")
print(net)
```

```r
# Cell 7 — Calculate Degree Centrality and Degree Distribution
node_degree <- degree(net, mode = "all")

degree_df <- data.frame(
  Node = names(node_degree),
  Degree = as.vector(node_degree)
)
degree_df <- degree_df[order(-degree_df$Degree), ]
row.names(degree_df) <- NULL

cat("Node Degree Centrality Table:\n")
print(degree_df)

par(mfrow = c(1, 1), mar = c(4, 4, 2, 1))
barplot(
  degree_df$Degree,
  names.arg = degree_df$Node,
  col = "steelblue",
  main = "Degree Distribution per Individual",
  ylab = "Degree (Number of Direct Connections)",
  las = 2
)
```

```r
# Cell 8 — Calculate Centrality Measures (Betweenness, Closeness, Eigenvector)
bet_centrality <- betweenness(net, directed = FALSE, normalized = TRUE)
close_centrality <- closeness(net, normalized = TRUE)
eigen_centrality_score <- eigen_centrality(net, directed = FALSE)$vector

centrality_table <- data.frame(
  Node = V(net)$name,
  Degree = as.vector(degree(net)),
  Betweenness = round(bet_centrality, 4),
  Closeness = round(close_centrality, 4),
  Eigenvector = round(eigen_centrality_score, 4)
)

centrality_table <- centrality_table[order(-centrality_table$Betweenness), ]
row.names(centrality_table) <- NULL

cat("Comprehensive Centrality Measures:\n")
print(centrality_table)
```

```r
# Cell 9 — Identify Most Influential Nodes
top_degree <- centrality_table[which.max(centrality_table$Degree), "Node"]
top_betweenness <- centrality_table[which.max(centrality_table$Betweenness), "Node"]
top_closeness <- centrality_table[which.max(centrality_table$Closeness), "Node"]
top_eigen <- centrality_table[which.max(centrality_table$Eigenvector), "Node"]

cat("================ KEY INFLUENTIAL ENTITIES ================\n")
cat("Most Directly Connected Entity (Hub - Degree)      :", top_degree, "\n")
cat("Primary Informational Gatekeeper (Betweenness)     :", top_betweenness, "\n")
cat("Fastest Network Communicator (Closeness)          :", top_closeness, "\n")
cat("Most Structurally Influential Entity (Eigenvector) :", top_eigen, "\n")
cat("==========================================================\n")
```

```r
# Cell 10 — Basic Network Visualization
set.seed(123)
plot(
  net,
  layout = layout_with_fr(net),
  main = "Basic Social Network Graph"
)
```

```r
# Cell 11 — Improved and Labelled Network Visualization
set.seed(123)
layout_coords <- layout_with_fr(net)

deg <- degree(net)
vertex_sizes <- 10 + (deg * 4)

plot(
  net,
  layout = layout_coords,
  vertex.size = vertex_sizes,
  vertex.color = "lightskyblue",
  vertex.frame.color = "darkblue",
  vertex.label = V(net)$name,
  vertex.label.cex = 0.9,
  vertex.label.color = "black",
  vertex.label.font = 2,
  edge.color = "gray60",
  edge.width = E(net)$weight * 0.8,
  main = "Improved Collaboration Network (Node Size ~ Degree, Edge Width ~ Tie Strength)"
)
legend(
  "bottomleft",
  legend = c("Node Size: Degree Connectivity", "Edge Width: Collaboration Intensity"),
  bty = "n",
  cex = 0.8
)
```

```r
# Cell 12 — Highlight Critical Nodes (Gatekeepers and Hubs)
set.seed(123)

node_colors <- rep("lightblue", vcount(net))
names(node_colors) <- V(net)$name

node_colors["Grace"] <- "coral"     # High Betweenness (Gatekeeper / Bridge)
node_colors["Frank"] <- "gold"      # High Eigenvector / Core Hub
node_colors["Jack"]  <- "gray85"    # Peripheral Node

plot(
  net,
  layout = layout_coords,
  vertex.size = 14 + (deg * 3),
  vertex.color = node_colors,
  vertex.frame.color = "black",
  vertex.label = V(net)$name,
  vertex.label.color = "black",
  vertex.label.font = 2,
  edge.color = "gray50",
  edge.width = 1.5,
  main = "Network Role Identification (Hub vs. Gatekeeper vs. Periphery)"
)
legend(
  "topleft",
  legend = c("Gatekeeper (Grace)", "Core Hub (Frank)", "Regular Member", "Periphery (Jack)"),
  fill = c("coral", "gold", "lightblue", "gray85"),
  bty = "n",
  cex = 0.85
)
```

```r
# Cell 13 — Community Detection (Clustering Analysis)
set.seed(123)
comm <- cluster_louvain(net)

cat("Number of detected communities:", length(comm), "\n")
cat("Modularity score:", round(modularity(comm), 4), "\n\n")

for (i in 1:length(comm)) {
  cat(paste0("Community ", i, ": "), paste(comm[[i]], collapse = ", "), "\n")
}

plot(
  comm,
  net,
  layout = layout_coords,
  vertex.size = 18,
  vertex.label.font = 2,
  main = "Community Detection via Louvain Modularity"
)
```
