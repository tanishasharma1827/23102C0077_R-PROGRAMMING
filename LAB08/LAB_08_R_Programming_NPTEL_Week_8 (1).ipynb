# PART A – LECTURE 31
## Operations on Lists

## Example 1 – Create Two Lists

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

list1
list2
```

## Example 2 – Merge Two Lists

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

list12 = c(list1, list2)
list12
```

## Example 3 – Convert a List to a Vector Using `unlist()`

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

unlist(list1)
unlist(list2)

mode(list1)
mode(unlist(list1))
```

## Example 4 – Append an Element to a List

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

append(list1, 100)
append(list2, "coffee")
```

## Example 5 – Append an Element After a Particular Position

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

## Example 6 – Remove an Element from a List

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

list1[-2]
list2[-1]
```

## Example 7 – Extract a Range of Elements from a Numeric List

```r
list1 = list(1, 2, 3, 4, 5, 6)

list1[2:4]
list1[c(1, 3, 5)]
```

## Example 8 – Extract a Range of Elements from a Character List

```r
list2 = list("water", "juice", "lemonade", "tea", "coffee", "milk")

list2[2:4]
list2[c(1, 3, 5)]
```

# PART B – LECTURE 32
## Vector Indexing

## Example 9 – Index a Vector Using a Condition

```r
x = 1:10
x
x[(x > 5)]
```

## Example 10 – Select Even and Odd Values Using Modulo

```r
x = 1:10

# %% indicates modulo
x[(x %% 2 == 0)]   # even values
x[(x %% 2 == 1)]   # odd values
```

## Example 11 – Handle Missing Observations with `NA`

```r
x = 1:10
x[5] = NA
x

y = x[!is.na(x)]
y

mean(x)
mean(y)
```

## Example 12 – Negative Integer Indexing

```r
x = 1:10
x

x[-(1:5)]
x[(6:10)]
```

## Example 13 – Named List and `names()`

```r
z = list(a1 = 1, a2 = "c", a3 = 1:3)

z
names(z)
```

## Example 14 – Change the Name of the Third List Element

```r
z = list(a1 = 1, a2 = "c", a3 = 1:3)

names(z)[3] = "c2"
z
```

## Example 15 – Named Vector and Access by Name

```r
x = c(water = 1, juice = 2, lemonade = 3)

names(x)
x["juice"]
```

## Example 16 – Empty Index

```r
x = 1:10

x
x[]
```

## Example 17 – Heterogeneous (Mixed-Mode) List with Dimensions

```r
ab = list(1, 2, 3, "X", "Y", "Z")
dim(ab) = c(2, 3)

print(ab)
mode(print(ab))
```

# PART C – LECTURE 33
## Factors

## Example 18 – Categorical Variable: Gender

```r
# Coding the categories as described in the notes:
# Male = 0, Female = 1

gender = c("Male", "Female", "Female", "Male", "Female")
gender_code = ifelse(gender == "Male", 0, 1)

gender
gender_code
```

## Example 19 – Categorical Variable: Performance

```r
# Performance categories and their numeric codes
performance = c("Excellent", "Average", "Good", "Bad")
performance_code = c(1, 2, 3, 4)

performance
performance_code
```

## Example 20 – Factor from Ball Colours

```r
# Red = 1, Blue = 2, Green = 3
ball_colours = c("Red", "Green", "Green", "Blue", "Red")

ball_codes = match(ball_colours, c("Red", "Blue", "Green"))

ball_colours
ball_codes
```

## Example 21 – Create a Factor with `factor()`

```r
x = c("Red", "Green", "Green", "Blue", "Red")
f = factor(x)

f
levels(f)
```

## Example 22 – Factor from a Die-Roll Example

```r
y = c(1, 4, 3, 5, 4, 2, 4)

possible.dieface = c(1, 2, 3, 4, 5, 6)
labels.dieface = c("one", "two", "three", "four", "five", "six")

facy = factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

facy
```

# PART D – LECTURE 34
## Factors – Class and Unclass

## Example 23 – Convert a Vector to a Factor Using `as.factor()`

```r
x = c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)

x
y = as.factor(x)
y
levels(y)
```

## Example 24 – Character Values Become Alphabetically Ordered Factor Levels

```r
x = factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water")
)

x
levels(x)
```

## Example 25 – Check the Class of Different R Objects

```r
class(9)
class("9")
class(print)

x = matrix(nrow = 2, ncol = 2, data = 1:4)
class(x)
x
```

## Example 26 – `unclass()` on Brand Factors

```r
brands = c("A", "A", "B", "B", "B", "B", "C")
brands

brands_fac = factor(brands)
brands_fac

unclass(brands_fac)
```

## Example 27 – Use Factor Codes to Index a Colour Vector

```r
colours = c("blue", "green", "red")
brands = c("A", "A", "B", "B", "B", "B", "C")
brands_fac = factor(brands)

unclass(brands_fac)
colours[unclass(brands_fac)]
```

## Example 28 – `unclass()` on a Factor with Alphabetical Levels

```r
x = factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water")
)

x
unclass(x)
```

## Example 29 – Specify a Different Order of Factor Levels

```r
x = factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water"),
  levels = c("water", "juice", "lemonade")
)

x
unclass(x)
levels(x)
```

## Example 30 – Ordered Factor

```r
income = ordered(
  c("high", "high", "low", "medium", "medium"),
  levels = c("low", "medium", "high")
)

income
unclass(income)
levels(income)
```
