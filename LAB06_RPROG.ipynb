## Section 1 — For Loop

The `for` loop is used when the number of repetitions is known in advance. Syntax:
```
for (name in vector) {
  commands to be executed
}
```


### Example 1 — Basic for loop

Loops over the integers 1 to 5 and prints the square of each value.

````
%%R
for ( i in 1:5 ) { print( i^2 ) }
````

### Example 2 — For loop over a custom vector

Loops over the vector `c(2,4,6,7)` and prints the square of each value. Here, `print` is a function used to print the argument.

````
%%R
for ( i in c(2,4,6,7) ) { print( i^2 ) }
````

### Example 3 — The `excount()` function

Demonstrates a `for` loop used inside a function. `excount()` counts how many elements of `x`, when divided by 2, are greater than 3.

````
%%R
x = c(2,4,6,8,10,12)

excount = function(x){
  count = 0
  for (xval in x) {
    if(xval/2 > 3)
      count = count+1
  }
  print(count)
}

excount(x)
````

## Section 2 — Nested For Loop

We can have a loop inside another loop. In this example, for every `child`, we loop through every `sweet` and print each combination.

````
%%R
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
````

## Section 3 — The `break` Command

Using the `break` command, we can stop the loop before it has looped through all the items. Here, the loop stops as soon as it encounters `"tea"`.

````
%%R
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}
````

## Section 4 — The `next` Command

Using the `next` command, we can skip an iteration without terminating the loop.

### Example — Skipping "lemonade"

````
%%R
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }
  print(x)
}
````

### Example — Skipping "tea"

````
%%R
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    next
  }
  print(x)
}
````

## Section 5 — While Loop

If the number of iterations is not known in advance (e.g. an iterative algorithm to maximize a likelihood function), a `while()` loop can be used.

Syntax:
```
while(condition){
  commands to be executed as long as condition is TRUE
}
```
If the condition is not true before entering the loop, no commands within the loop are executed.

### While Loop Example 1

Prints the square of `i`, starting at 1 and incrementing by 2, as long as `i < 10`.

````
%%R
i = 1

while (i < 10) {
  print(i^2)
  i = i+2
}
````

### While Loop Example 2 — `sumfunction()`

**Note on modification:** The original PDF example uses `readline()` to interactively ask the user to enter a number less than 25 (`number = as.integer(readline(prompt="Please select any number less than 25: "))`). Interactive `readline()` input is not practical in a Google Colab code cell (there is no interactive console prompt during batch execution), so the input has been replaced with a fixed value, `number = 22`, matching the exact value used in the PDF's own demonstration output. The logic of the function (looping with `while` and accumulating the sum) is otherwise preserved exactly as in the PDF.

````
%%R
sumfunction = function(){
  sum = 0
  number = 22

  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }

  print(paste(
    "The sum of numbers received from the While Loop: ",
    sum
  ))
}

sumfunction()
````

## Section 6 — Repeat Loop

The `repeat` loop does not test any condition before entering the loop, and it also does not terminate the execution of the loop on its own. The programmer is responsible for ensuring the loop terminates after the appropriate number of iterations, for which the `break` command is used.

Syntax:
```
repeat{
  commands to be executed
}
```

### Repeat Loop Example 1

Prints the square of `i` (starting at 1, incrementing by 2) and stops using `break` once `i > 10`.

````
%%R
i = 1

repeat{
  print(i^2)
  i = i+2

  if(i > 10)
    break
}
````

### Repeat Loop Example 2 — Using both `next` and `break`

The `next` command returns to the beginning of the loop (the first command in the loop) without executing the remaining commands for that iteration.

````
%%R
i = 1

repeat{
  i = i+1

  if(i < 10)
    next

  print(i^2)

  if(i >= 13)
    break
}
````

## Section 7 — Functions

Functions are a bunch of commands grouped together as a sensible unit. Functions take input arguments, perform calculations (or produce graphics, call other functions), and produce some output, usually returning a result. A function consists of four main components:

* **Function Name** — the name of the function, stored as an object with this name.
* **Arguments** — the input values; a function may contain no arguments.
* **Function Body** — statements that define what the function has to do.
* **Return Value** — the evaluated value of the last expression in the function body.

### Function Example 1 — Single Variable

Defines a function `abc(x)` that returns `x^2`, and calls it with different arguments.

````
%%R
abc = function(x){
  x^2
}

abc(3)
abc(6)
abc(9)
````

### Function Example 2 — Two Variables

Defines a function `abc(x,y)` that returns `x^2 + y^2`.

````
%%R
abc = function(x,y){
  x^2+y^2
}

abc(3,4)
abc(10,10)
abc(-2,-3)
````

### Function Example 3 — Mathematical/Trigonometric Function

Defines a function `abc(x)` that returns `sin(x)^2 + cos(x)^2 + x`.

````
%%R
abc = function(x){
  sin(x)^2+cos(x)^2+x
}

abc(9)
abc(99)
abc(-15)
````

### Function Example 4 — Function Without Argument

Demonstrates calling a function that takes no arguments. The function `abc()` uses a `for` loop internally to print the cube of `i` for `i` in `1:3`.

````
%%R
abc = function() {
  for(i in 1:3) {
    print(i^3)
  }
}

abc()
````

## Section 8 — Sequences

A sequence is a set of related numbers, events, movements, or items that follow each other in a particular order. Regular sequences can be generated in R using the `seq()` function.

Syntax:
```
seq(from = 1, to = 1, by = ((to - from)/(length.out - 1)),
    length.out = NULL, along.with = NULL, ...)
```

### Sequence Example 1

The default increment is +1. Generates a sequence from 2 to 4.

````
%%R
seq(from=2, to=4)
````

### Sequence Example 2

The default increment is −1 when `from` > `to`. Generates a sequence from 4 to 2.

````
%%R
seq(from=4, to=2)
````

### Sequence Example 3

Generates a sequence from −4 to 4 with the default increment of +1.

````
%%R
seq(from=-4, to=4)
````

### Sequence Example 4 — Constant Increment

Generates a sequence from 10 to 20 with an increment of 2 units.

````
%%R
seq(from=10, to=20, by=2)
````

### Sequence Example 5 — Constant Decrement

Generates a sequence from 20 to 10 with a decrement of 2 units.

````
%%R
seq(from=20, to=10, by=-2)
````

### Sequence Example 6 — Fractional Decrement

Generates a sequence from 3 to −2 with a decrement of 0.5 units.

````
%%R
seq(from=3, to=-2, by=-0.5)
````

### Sequence Example 7 — Predefined Length

Generates a sequence of length 10 ending at 10, using the default increment of +1.

````
%%R
seq(to=10, length=10)
````

### Sequence Example 8 — Predefined Length Starting From a Value

Generates a sequence of length 10 starting at 10, using the default increment of +1.

````
%%R
seq(from=10, length=10)
````

### Sequence Example 9 — Fractional Increment with Length

Generates a sequence of length 10 starting at 10, with a fractional increment of 0.1.

````
%%R
seq(from=10, length=10, by=0.1)
````

### Sequence Example 10 — Decrement with Length

Generates a sequence of length 10 starting at 10, with a decrement of 2.

````
%%R
seq(from=10, length=10, by=-2)
````

### Sequence Example 11 — Fractional Decrement with Length

Generates a sequence of length 5 starting at 10, with a fractional decrement of 0.2.

````
%%R
seq(from=10, length=5, by=-.2)
````
